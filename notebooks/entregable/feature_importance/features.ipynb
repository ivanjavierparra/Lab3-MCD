```python
import json
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Leer el archivo JSON
with open("./v20.json", "r") as f:
    data = json.load(f)

# Convertir a DataFrame
df_importancia = pd.DataFrame(list(data.items()), columns=["feature", "importance"])
df_importancia
```

```python
df_top30 = df_importancia.sort_values("importance", ascending=False).head(30)
```

```python
plt.figure(figsize=(10, 8))
plt.barh(df_top30["feature"], df_top30["importance"], color='skyblue')
plt.gca().invert_yaxis()  # La más importante arriba
plt.xlabel("Importancia")
plt.title("Top 30 Variables Más Importantes")
plt.tight_layout()
plt.show()
```
